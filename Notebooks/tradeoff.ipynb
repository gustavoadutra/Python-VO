```python
import matplotlib.pyplot as plt

# 1. Configurações de estilo acadêmico (IEEE / SBC / Elsevier)
plt.rcParams.update({
    'font.family': 'serif',       # Fonte clássica com serifa
    'font.size': 11,              # Tamanho base para leitura em duas colunas
    'axes.labelsize': 12,         # Tamanho dos rótulos (X e Y)
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'legend.fontsize': 10,
    'text.usetex': False          # Deixe True se tiver o LaTeX (texlive) instalado na máquina
})

# 2. Dados extraídos do seu dataset (Tempo em ms vs Erro ATE VO em metros)
# Formato: 'Config': (Tempo_Total_Segundos * 1000, ATE_VO, Marcador, Cor)
data = {
    'ORB + KNN': (0.01828 * 1000, 0.0966, 'o', '#4472C4'),
    'SIFT + FLANN': (0.03335 * 1000, 0.0905, 's', '#ED7D31'),
    'SIFT + LIGHT': (0.09983 * 1000, 0.1005, 'D', '#A5A5A5'),
    'SUPERPOINT + FLANN': (0.33502 * 1000, 0.1066, '^', '#70AD47'),
    'SUPERPOINT + LIGHT': (0.39172 * 1000, 0.1170, 'v', '#9E480E'),
    'XFEAT + FLANN': (0.06241 * 1000, 0.1359, 'p', '#FFC000'),
    'XFEAT + LIGHT': (0.26377 * 1000, 0.0903, '*', '#FF0000')  # Destacado por obter o menor erro
}

fig, ax = plt.subplots(figsize=(6.5, 4.5), dpi=300)

# 3. Plotagem dos pontos puros (Scatter)
for label, (time_ms, error_m, marker, color) in data.items():
    # Aumenta ligeiramente o tamanho da estrela do XFEAT+LIGHT para dar destaque
    size = 12 if 'LIGHT' in label and 'XFEAT' in label else 8
    
    ax.scatter(
        time_ms, error_m, 
        marker=marker, 
        color=color, 
        s=size**2, 
        label=label, 
        alpha=0.9, 
        edgecolors='black',  # Linha preta fina no contorno para destaque acadêmico
        linewidths=0.5
    )

# 4. Rótulos formais dos eixos
ax.set_xlabel('Average Total Time per Frame (ms)')
ax.set_ylabel('Absolute Trajectory Error - ATE VO (m)')

# 5. Estética da Moldura e Grid
ax.spines['top'].set_visible(False)     # Remove linha superior
ax.spines['right'].set_visible(False)   # Remove linha direita
ax.grid(True, linestyle='--', alpha=0.5, which='both') # Grid discreto

# Posicionamento da Legenda
ax.legend(loc='upper right', frameon=True, facecolor='white', edgecolor='none')

plt.tight_layout()

# Salva em PDF (vetorial) para não perder resolução ao dar zoom no artigo
plt.savefig('cuscobot_performance_tradeoff.pdf', bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Configurações de estilo acadêmico
plt.rcParams.update({
    'font.family': 'serif',       # Mantém o padrão de fontes clássicas para artigos
    'font.size': 11,
    'axes.labelsize': 12,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'legend.fontsize': 10,
    'text.usetex': False          # Altere para True se for compilar diretamente com LaTeX local
})

# 2. Dados mapeando o Tempo Total (ms) vs RPE VO (m)
# Formato: 'Config': (Tempo_Total_Segundos * 1000, RPE_VO, Marcador, Cor)
data_rpe = {
    'ORB + KNN': (0.01828 * 1000, 0.1128, 'o', '#4472C4'),
    'SIFT + FLANN': (0.03335 * 1000, 0.1110, 's', '#ED7D31'),
    'SIFT + LIGHT': (0.09983 * 1000, 0.1061, 'D', '#A5A5A5'),
    'SUPERPOINT + FLANN': (0.33502 * 1000, 0.1098, '^', '#70AD47'),
    'SUPERPOINT + LIGHT': (0.39172 * 1000, 0.1140, 'v', '#9E480E'),
    'XFEAT + FLANN': (0.06241 * 1000, 0.1214, 'p', '#FFC000'),
    'XFEAT + LIGHT': (0.26377 * 1000, 0.1051, '*', '#FF0000')  # Destacado em vermelho por apresentar menor erro
}

fig, ax = plt.subplots(figsize=(6.5, 4.5), dpi=300)

# 3. Plotagem dos pontos (Scatter)
for label, (time_ms, error_m, marker, color) in data_rpe.items():
    # Destaca o tamanho do marcador XFEAT + LIGHT
    size = 12 if 'LIGHT' in label and 'XFEAT' in label else 8
    
    ax.scatter(
        time_ms, error_m, 
        marker=marker, 
        color=color, 
        s=size**2, 
        label=label, 
        alpha=0.9, 
        edgecolors='black', 
        linewidths=0.5
    )

# 4. Ajustes formais de eixos (RPE em metros)
ax.set_xlabel('Average Total Time per Frame (ms)')
ax.set_ylabel('Relative Pose Error - RPE VO (m)')

# 5. Limpeza de bordas e Grid Acadêmico
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(True, linestyle='--', alpha=0.5, which='both')

# Posicionamento da Legenda
ax.legend(loc='upper right', frameon=True, facecolor='white', edgecolor='none')

plt.tight_layout()

# Salvando em formato PDF vetorial para garantir a qualidade de impressão do artigo
plt.savefig('cuscobot_rpe_tradeoff.pdf', bbox_inches='tight')
plt.show()
```
